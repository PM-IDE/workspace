```python
import pm4py
import os
from ficus import *
import pandas as pd

filtering_parts = [
    RemainEventsByRegex('GC/'),
    FilterEventsByRegex('GC/RestartEEStart'),
    FilterEventsByRegex('GC/RestartEEStop'),
    FilterEventsByRegex('GC/SuspendEEStart'),
    FilterEventsByRegex('GC/SuspendEEStop'),
    FilterEventsByRegex('GC/Finaliz'),
    FilterEventsByRegex('GC/SampledObject'),
    FilterEventsByRegex('GC/SetGCHandle'),
    FilterEventsByRegex('GC/BGCAllocWait'),
    FilterEventsByRegex('GC/Pin'),
    FilterEventsByRegex('GC/CreateSegment'),
    FilterEventsByRegex('GC/Triggered'),
]

def calculate_metrics(log, p, i, e):
    fitness = pm4py.fitness_token_based_replay(log, p, i, e)['log_fitness']
    precision = pm4py.precision_token_based_replay(log, p, i, e)
    simplicity = pm4py.simplicity_petri_net(p, i, e)
    generalization = pm4py.generalization_tbr(log, p, i, e)
    return [fitness, precision, simplicity, generalization]

def create_metrics_df():
    log = pm4py.read_xes('gc_log.xes', return_legacy_log_object=True)
    algorithms = [
        ('inductive', pm4py.discover_petri_net_inductive),
        ('heuristic', pm4py.discover_petri_net_heuristics),
        ('alpha', pm4py.discover_petri_net_alpha),
        ('alpha_plus', pm4py.discover_petri_net_alpha_plus),
    ]

    metrics = []
    for (algo, func) in algorithms:
        p, i, e = func(log)
        metrics.append([algo] + calculate_metrics(log, p, i, e))

    p, i, e = pm4py.read_pnml('original_ecfg.pnml')
    p = pm4py.reduce_petri_net_invisibles(p)

    metrics.append(['ecfg'] + calculate_metrics(log, p, i, e))

    return pd.DataFrame(metrics, columns=['algo', 'fitness', 'precision', 'simplicity', 'generalization'])

log_name = 'concurrent_server_1.bxes'
log_bytes = BytesContextValue(read_file_bytes(os.path.join(os.path.abspath(os.curdir), log_name)))
```

```python
Pipeline(
    ReadLogFromBxes(use_bytes=True),
    filtering_parts,
    DiscoverCases(
        start_regex='GC/Start',
        end_regex='GC/Stop',
        inline_inner_cases=True,
        pipeline=Pipeline(
            AddStartArtificialEvents(),

            WriteLogToXesBytes(save_path=os.path.join(os.path.abspath(os.curdir), 'gc_log.xes')),
            DiscoverECFG(root_sequence_kind=RootSequenceKind.FindBest, merge_sequences_of_events=False),
            ConvertECFGToPetriNet(),
            SerializePetriNetPnmlBytes(save_path=os.path.join(os.path.abspath(os.curdir), 'original_ecfg.pnml')),
            GetPetriNetInfo(),

            DiscoverMultithreadedDfg(thread_attribute='NativeThreadId'),
            ViewGraph(export_path=os.path.join(os.path.abspath(os.curdir), 'multithreaded.png')),
            GetGraphInfo(),
            SerializeGraphPromBytes(save_path=os.path.join(os.path.abspath(os.curdir), 'multithreaded.dfg')),
            DiscoverDirectlyFollowsGraph(),
            ViewGraph(export_path=os.path.join(os.path.abspath(os.curdir), 'default.png')),
            SerializeGraphPromBytes(save_path=os.path.join(os.path.abspath(os.curdir), 'default.dfg')),
        )
    )
).execute(
    'localhost:1234',
    initial_context={
        'bytes': log_bytes,
    }
)

create_metrics_df()
```

```python
def calculate_metrics_for_dfg_nets():
    log = pm4py.read_xes('gc_log.xes', return_legacy_log_object=True)

    metrics = []
    for net in ['prom_default.pnml', 'prom_mthreaded.pnml']:
        p, i, e = pm4py.read_pnml(net)
        metrics.append([net] + calculate_metrics(log, p, i, e))

    return metrics

pd.DataFrame(calculate_metrics_for_dfg_nets(), columns=['net', 'fitness', 'precision', 'simplicity', 'generalization'])
```

```python
Pipeline(
    ReadLogFromBxes(use_bytes=True),
    filtering_parts,
    DiscoverCases(
        start_regex='GC/Start',
        end_regex='GC/Stop',
        inline_inner_cases=True,
        pipeline=Pipeline(
            AbstractMultithreadedEventsGroups(
                thread_attribute='NativeThreadId',
                time_attribute='QpcStamp',
                n_components=NComponents.Two,
                distance=Distance.Cosine,
                tolerance=0.1,
                view_params=(-50, 20),
                put_noise_events_in_one_cluster=False,
                min_events_count_in_cluster=2,
                feature_count_kind=FeatureCountKind.Count,
                visualization_method=DatasetVisualizationMethod.TSNE,
                after_clusterization_pipeline=Pipeline(
                    AddStartArtificialEvents(),

                    DiscoverLoopsStrict(max_loop_length=100),
                    CreateLogFromActivitiesInstances(strategy=UndefinedActivityHandlingStrategy.InsertAllEvents),

                    WriteLogToXesBytes(save_path=os.path.join(os.path.abspath(os.curdir), 'gc_log.xes')),
                    DiscoverECFG(root_sequence_kind=RootSequenceKind.FindBest, merge_sequences_of_events=False),
                    ViewGraph(),
                    ConvertECFGToPetriNet(),
                    SerializePetriNetPnmlBytes(
                        save_path=os.path.join(os.path.abspath(os.curdir), 'original_ecfg.pnml')),
                )
            )
        )
    )
).execute(
    'localhost:1234',
    initial_context={
        'bytes': log_bytes,
    }
)

create_metrics_df()
```
