```python
import pm4py
import os
from ficus import *
import pandas as pd

root_sequence_kind = RootSequenceKind.FindBest
filtering_parts = [
    RemainEventsByRegex('GC/'),
    FilterEventsByRegex('GC/RestartEEStart'),
    FilterEventsByRegex('GC/RestartEEStop'),
    FilterEventsByRegex('GC/SuspendEEStart'),
    FilterEventsByRegex('GC/SuspendEEStop'),
    FilterEventsByRegex('GC/Finaliz'),
    FilterEventsByRegex('GC/SampledObject'),
    FilterEventsByRegex('GC/SetGCHandle'),
    FilterEventsByRegex('GC/BGCAllocWait'),
    FilterEventsByRegex('GC/Pin'),
    FilterEventsByRegex('GC/CreateSegment'),
    FilterEventsByRegex('GC/Triggered'),
]

def path(name: str) -> str:
    return os.path.join(os.path.abspath(os.curdir), name)

def calculate_metrics(log, p, i, e):
    fitness = pm4py.fitness_token_based_replay(log, p, i, e)['log_fitness']
    precision = pm4py.precision_token_based_replay(log, p, i, e)
    simplicity = pm4py.simplicity_petri_net(p, i, e)
    generalization = pm4py.generalization_tbr(log, p, i, e)
    return [fitness, precision, simplicity, generalization]

def create_metrics_df(exp_num: int):
    log = pm4py.read_xes(f'gc_log_{exp_num}.xes', return_legacy_log_object=True)
    algorithms = [
        ('inductive', pm4py.discover_petri_net_inductive),
        ('heuristic', pm4py.discover_petri_net_heuristics),
        ('alpha', pm4py.discover_petri_net_alpha),
        ('alpha_plus', pm4py.discover_petri_net_alpha_plus),
    ]

    metrics = []
    for (algo, func) in algorithms:
        p, i, e = func(log)
        petri_net_info = [len(p.places), len(p.transitions), len(p.arcs)]
        pm4py.save_vis_petri_net(p, i, e, path(f'exp_{exp_num}_{algo}.png'))
        metrics.append([algo] + calculate_metrics(log, p, i, e) + petri_net_info)

    p, i, e = pm4py.read_pnml(f'ecfg_{exp_num}.pnml')
    p = pm4py.reduce_petri_net_invisibles(p)
    pm4py.save_vis_petri_net(p, i, e, path(f'exp_{exp_num}_ecfg.png'))
    petri_net_info = [len(p.places), len(p.transitions), len(p.arcs)]
    metrics.append(['ecfg'] + calculate_metrics(log, p, i, e) + petri_net_info)

    return pd.DataFrame(metrics, columns=['algo', 'fitness', 'precision', 'simplicity', 'generalization', 'places', 'transitions', 'arcs'])

log_name = 'concurrent_server_1.bxes'
log_bytes = BytesContextValue(read_file_bytes(path(log_name)))
```

```python
Pipeline(
    ReadLogFromBxes(use_bytes=True),
    filtering_parts,
    DiscoverCases(
        start_regex='GC/Start',
        end_regex='GC/Stop',
        inline_inner_cases=True,
        pipeline=Pipeline(
            AddStartArtificialEvents(),

            TracesDiversityDiagram(save_path=path('exp1.png'), height_scale=20, width_scale=10),
            WriteLogToXesBytes(save_path=path('gc_log_1.xes')),
            PrintEventLogInfo(),
            DiscoverECFG(root_sequence_kind=root_sequence_kind, merge_sequences_of_events=False),
            GetGraphInfo(),
            ConvertECFGToPetriNet(),
            SerializePetriNetPnmlBytes(save_path=path('ecfg_1.pnml')),
            GetPetriNetInfo(),

            DiscoverMultithreadedDfg(thread_attribute='NativeThreadId'),
            ViewGraph(export_path=path('multithreaded.png')),
            SerializeGraphPromBytes(save_path=path('multithreaded.dfg')),
            DiscoverDirectlyFollowsGraph(),
            ViewGraph(export_path=path('default.png')),
            SerializeGraphPromBytes(save_path=path('default.dfg')),
        )
    )
).execute(
    'localhost:1234',
    initial_context={
        'bytes': log_bytes,
    }
)

create_metrics_df(1)
```

```python
def calculate_metrics_for_dfg_nets():
    log = pm4py.read_xes('gc_log_1.xes', return_legacy_log_object=True)

    metrics = []
    for net in ['prom_default.pnml', 'prom_mthreaded.pnml']:
        p, i, e = pm4py.read_pnml(net)
        petri_net_info = [len(p.places), len(p.transitions), len(p.arcs)]
        metrics.append([net] + calculate_metrics(log, p, i, e) + petri_net_info)

    return metrics

pd.DataFrame(calculate_metrics_for_dfg_nets(), columns=['algo', 'fitness', 'precision', 'simplicity', 'generalization', 'places', 'transitions', 'arcs'])
```
